```python
import pandas as pd
import xgboost as xgb
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder
import joblib
import os
```

```python
# Load res_parsed
train_df = pd.read_csv('res_parsed/train.csv')
val_df = pd.read_csv('res_parsed/val.csv')
test_df = pd.read_csv('res_parsed/test.csv')

# Features to use
features = ['mean_h', 'mean_s', 'mean_v', 'std_h', 'std_s', 'std_v']

X_train = train_df[features]
y_train = train_df['label']

X_val = val_df[features]
y_val = val_df['label']

X_test = test_df[features]
y_test = test_df['label']

# Encode string labels
label_encoder = LabelEncoder()
y_train_enc = label_encoder.fit_transform(y_train)
y_val_enc = label_encoder.transform(y_val)
y_test_enc = label_encoder.transform(y_test)
```

```python
model = xgb.XGBClassifier(
    n_estimators=150,
    max_depth=5,
    learning_rate=0.1,
    eval_metric='mlogloss',
    random_state=42
)

model.fit(
    X_train, y_train_enc,
    eval_set=[(X_val, y_val_enc)],
    verbose=False
)

```

```python
y_pred = model.predict(X_test)

print("🧪 Test Accuracy:", accuracy_score(y_test_enc, y_pred))
print(classification_report(y_test_enc, y_pred, target_names=label_encoder.classes_))
print(confusion_matrix(y_test_enc, y_pred))
```

```python
os.makedirs("model", exist_ok=True)
joblib.dump(model, "model/xgboost_color_classifier.pkl")
joblib.dump(label_encoder, "model/label_encoder.pkl")
print("XGBoost model saved.")
```
